```python
import pandas as pd
import glob
```

```python
# ----------------------------
# Load transaction files
# ----------------------------
transaction_files = glob.glob(
    r"D:\DataScience\GUVI\Local\Telangana_govt_Ration_distribution_clustering\Data\shop-wise-trans-details_*_*.csv"
)

print(f"Transaction files found: {len(transaction_files)}")
for file in transaction_files:
    print(file)

transactions = pd.concat(
    [pd.read_csv(file) for file in transaction_files],
    ignore_index=True
)

print(f"\nTotal transaction records: {len(transactions):,}")

# ----------------------------
# Load card status files
# ----------------------------
card_status_files = glob.glob(
    r"D:\DataScience\GUVI\Local\Telangana_govt_Ration_distribution_clustering\Data\fpshop-card-status_*_*.csv"
)

print(f"\nCard Status files found: {len(card_status_files)}")
for file in card_status_files:
    print(file)

card_status = pd.concat(
    [pd.read_csv(file) for file in card_status_files],
    ignore_index=True
)

print(f"\nTotal card status records: {len(card_status):,}")
shop_details=pd.read_csv(r"D:\DataScience\GUVI\Local\Telangana_govt_Ration_distribution_clustering\Data\shop-status-details_6_2025.csv")


# ----------------------------
# Triple Join
# ----------------------------
master_df = (
    transactions
    .merge(card_status, on=["shopNo", "distCode", "month", "year"], how="left")
    .merge(shop_details, on=["shopNo", "distCode"], how="left")
)

print(f"\nMaster dataset records: {len(master_df):,}")

# Save
master_df.to_csv("master_dataset.csv", index=False)

print("\nFirst 5 rows:")
print(master_df.head())
```
